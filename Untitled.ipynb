```python
import pandas as pd
import numpy as np

df["Other groups"] = df["Other groups"].fillna('None')

df["is_vital_registration_complete"]=df["Vital registration complete"].notnull().astype(int)
df=df.drop(columns=['Vital registration complete'])

df['has_special_notes'] = df['Special Notes'].notnull().astype(int)
df = df.drop(columns=['Special Notes'])

df = df.drop(columns=['National accounts reference year'])

redundant_cols = ['2-alpha code', 'WB-2 code', 'Table Name']
df = df.drop(columns=[col for col in redundant_cols if col in df.columns])
```

```python
# Clean all column names: lowercase, trimmed, underscores instead of spaces/slashes
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r'[\s/]+', '_', regex=True)
    .str.replace(r'[^\w_]', '', regex=True)  # removes special symbols like hyphens
)

print(df.columns.tolist()[:5])
# Output: ['code', 'long_name', 'income_group', 'region', 'lending_category']
```

```python
# Extract the first 4-digit number, then convert to Int64
df['national_accounts_base_year'] = (
    df['national_accounts_base_year']
    .astype(str)
    .str.extract(r'(\b\d{4}\b)')[0]
    .pipe(pd.to_numeric, errors='coerce')
    .astype('Int64')
)
```

```python
# 1. Calculate the mean and round to the nearest whole year
mean_year = round(df['national_accounts_base_year'].mean())

# 2. Fill missing values with the rounded mean
df['national_accounts_base_year'] = df['national_accounts_base_year'].fillna(mean_year).astype('Int64')
```

```python
cols_to_convert = [
    'latest_population_census',
    'latest_industrial_data',
    'latest_trade_data',
    'latest_water_withdrawal_data'
]

df[cols_to_convert] = df[cols_to_convert].apply(pd.to_numeric, errors='coerce').astype('Int64')
```

```python
import glob
import os
import sqlite3
import pandas as pd

# 1. Create an in-memory SQLite database
conn = sqlite3.connect(":memory:")

# 2. Find all CSV files in the current directory (or list your 5 filenames explicitly)
csv_files = glob.glob("*.csv")
# Alternatively: csv_files = ['file1.csv', 'file2.csv', 'file3.csv', 'file4.csv', 'file5.csv']

for file in csv_files:
  # Clean the filename to make a valid SQL table name
  table_name = os.path.splitext(os.path.basename(file))[0]
  table_name = "".join(c if c.isalnum() else "_" for c in table_name).lower()

  # Read CSV and clean column names (replace spaces/special chars with underscores)
  df = pd.read_csv(file)   
  df.columns = [
      "".join(c if c.isalnum() else "_" for c in str(col)).lower()
      for col in df.columns
  ]

  # Write DataFrame to the SQLite database
  df.to_sql(table_name, conn, if_exists="replace", index=False)
  print(f"Loaded {file} -> table '{table_name}' ({len(df)} rows)")

# 3. Export the entire database (schema + inserts) to a single .sql file
output_sql_file = "combined_datasets.sql"
with open(output_sql_file, "w", encoding="utf-8") as f:
  for line in conn.iterdump():
    f.write(f"{line}\n")

conn.close()
print(f"\nSuccess! Exported all tables to {output_sql_file}")
```

```python
import pandas as pd

# ==========================================================
# 1. CLEAN: 01_country_metadata_cleaned.csv
# ==========================================================
f1 = "data/processed/01_country_metadata_cleaned.csv"
df_countries = pd.read_csv(f1)

# Safely rename 'code' -> 'country_code' only if 'code' still exists
if "code" in df_countries.columns:
    df_countries = df_countries.rename(columns={"code": "country_code"})
df_countries["country_code"] = df_countries["country_code"].str.strip()

keep_country_cols = [
    "country_code",
    "short_name",
    "long_name",
    "region",
    "income_group",
    "lending_category",
    "currency_unit",
    "national_accounts_base_year",
    "external_debt_reporting_status",
    "latest_population_census",
    "latest_industrial_data",
    "latest_trade_data",
    "latest_water_withdrawal_data",
    "is_vital_registration_complete",
    "has_special_notes",
]
df_countries = df_countries[
    [c for c in keep_country_cols if c in df_countries.columns]
].drop_duplicates(subset=["country_code"])

df_countries["income_group"] = df_countries["income_group"].fillna(
    "Low income"
)
year_meta_cols = [
    "latest_population_census",
    "latest_industrial_data",
    "latest_trade_data",
    "latest_water_withdrawal_data",
]
df_countries[year_meta_cols] = df_countries[year_meta_cols].astype("Int64")

df_countries.to_csv(f1, index=False)
print(f"Updated {f1}: {df_countries.shape}")

# ==========================================================
# 2. CLEAN: 05_IDS_SeriesMetaData_cleaned.csv
# ==========================================================
f5 = "data/processed/05_IDS_SeriesMetaData_cleaned.csv"
df_indicators = pd.read_csv(f5)

if "code" in df_indicators.columns:
    df_indicators = df_indicators.rename(columns={"code": "series_code"})
df_indicators["series_code"] = df_indicators["series_code"].str.strip()

keep_indicator_cols = [
    "series_code",
    "indicator_name",
    "long_definition",
    "topic",
    "source",
    "dataset",
    "periodicity",
    "aggregation_method",
]
df_indicators = df_indicators[
    [c for c in keep_indicator_cols if c in df_indicators.columns]
].drop_duplicates(subset=["series_code"])
df_indicators["dataset"] = df_indicators["dataset"].fillna(
    "International Debt Statistics"
)

df_indicators.to_csv(f5, index=False)
print(f"Updated {f5}: {df_indicators.shape}")

# Master valid key sets
valid_countries = set(df_countries["country_code"])
valid_indicators = set(df_indicators["series_code"])

# ==========================================================
# 3. CLEAN: 03_IDS_ALLCountries_Data_cleaned.csv
# ==========================================================
f3 = "data/processed/03_IDS_ALLCountries_Data_cleaned.csv"
df_debt = pd.read_csv(f3)

df_debt["country_code"] = df_debt["country_code"].str.strip()
df_debt["series_code"] = df_debt["series_code"].str.strip()

unwanted_debt_cols = [
    "country_name",
    "series_name",
    "counterpartarea_name",
    "counterpartarea_code",
]
df_debt = df_debt.drop(columns=unwanted_debt_cols, errors="ignore")

df_debt = df_debt[
    df_debt["country_code"].isin(valid_countries)
    & df_debt["series_code"].isin(valid_indicators)
].drop_duplicates(subset=["country_code", "series_code"])

df_debt.to_csv(f3, index=False)
print(f"Updated {f3}: {df_debt.shape}")

# ==========================================================
# 4. CLEAN: 02_FootNoteMetaData_cleaned.csv
# ==========================================================
f2 = "data/processed/02_FootNoteMetaData_cleaned.csv"
df_fn = pd.read_csv(f2)

df_fn["country_code"] = df_fn["country_code"].str.strip()
df_fn["series_code"] = df_fn["series_code"].str.strip()

if "time_code" in df_fn.columns:
    df_fn["year"] = (
        df_fn["time_code"].str.replace("YR", "", regex=False).astype(int)
    )
df_fn = df_fn.drop(columns=["type", "time_code"], errors="ignore")

df_fn = df_fn[
    df_fn["country_code"].isin(valid_countries)
    & df_fn["series_code"].isin(valid_indicators)
][["country_code", "series_code", "year", "description"]].drop_duplicates(
    subset=["country_code", "series_code", "year"]
)

df_fn.to_csv(f2, index=False)
print(f"Updated {f2}: {df_fn.shape}")

# ==========================================================
# 5. CLEAN: 04_Country_Series_Metadata_cleaned.csv
# ==========================================================
f4 = "data/processed/04_Country_Series_Metadata_cleaned.csv"
df_csm = pd.read_csv(f4)

df_csm["country_code"] = df_csm["country_code"].str.strip()
df_csm["series_code"] = df_csm["series_code"].str.strip()

df_csm = df_csm[
    df_csm["country_code"].isin(valid_countries)
    & df_csm["series_code"].isin(valid_indicators)
].drop_duplicates(subset=["country_code", "series_code"])

df_csm.to_csv(f4, index=False)
print(f"Updated {f4}: {df_csm.shape}")
```
